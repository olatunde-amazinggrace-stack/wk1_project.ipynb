**Covid 19 Histogram Chart**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('/content/time_series_covid_19_confirmed.csv')

# =========================
# 1. HISTOGRAM
# =========================
plt.figure(figsize=(8,5))
plt.hist(df.iloc[:, -1], bins=20)

plt.title('Distribution of Confirmed Cases')
plt.xlabel('Confirmed Cases')
plt.ylabel('Frequency')

plt.show()
```

**Covid 19 Scatter Plot**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset for confirmed cases
df = pd.read_csv('/content/time_series_covid_19_confirmed.csv')
# Load dataset for deaths
df_deaths = pd.read_csv('/content/time_series_covid_19_deaths.csv')

# =========================
# 2. SCATTER PLOT
# =========================
plt.figure(figsize=(8,5))
# Use the last column of df for confirmed cases (x-axis)
# Use the last column of df_deaths for deaths (y-axis)
plt.scatter(df.iloc[:, -1], df_deaths.iloc[:, -1])

plt.title('Confirmed Cases vs Deaths')
plt.xlabel('Confirmed Cases (Latest Date)')
plt.ylabel('Deaths (Latest Date)')

plt.show()
```

**Covid 19 Bar Chart**

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv('/content/time_series_covid_19_confirmed.csv')

# =========================
# 3. BAR CHART
# =========================
country_cases = df.groupby('Country/Region').sum().iloc[:, -1]

plt.figure(figsize=(10,5))
country_cases.plot(kind='bar')

plt.title('Top 10 Countries by Confirmed Cases')
plt.xlabel('Country')
plt.ylabel('Confirmed Cases')

plt.show()
```

**Covid 19 Box Plot**
