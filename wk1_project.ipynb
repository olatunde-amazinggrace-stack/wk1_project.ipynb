TASK 1 — AI Co-pilot Loop

The AI co-pilot loop is the process of working together with an AI tool to solve problems, especially in coding and data analysis. It is usually done step by step until the desired result is achieved.

1. Prompt

The first step is giving instructions or asking the AI a question. This is called a prompt.
For example, a user may ask the AI to write Python code to analyze a dataset.

2. Generate Code

After receiving the prompt, the AI generates code or a possible solution based on the instruction given.

3. Run Code

The generated code is then executed in a programming environment such as Jupyter Notebook, Google Colab, or VS Code to see if it works correctly.

4. Read Errors/Results

Once the code runs, the user checks the output. If there are errors, the user studies the error messages. If the code works, the results are examined to confirm they are correct and meaningful.

5. Fix/Improve

Based on the errors or results, the user may ask the AI to correct mistakes, improve efficiency, add features, or explain unclear parts of the code.

6. Repeat

The process continues repeatedly until the code works properly and produces the expected outcome. This continuous interaction between the user and the AI is called the AI co-pilot loop.

```python
import pandas as pd
diabetes_df = pd.read_csv("/diabetes.csv")
```

```python
# Display the first 5 rows
print(diabetes_df.head())
```

```python
# Display information about the dataset
print(diabetes_df.info())
```

```python
# Display statistical summary
print(diabetes_df.describe())
```

```python
# Count how many patients have diabetes
# Assuming the Outcome column:
# 1 = Diabetic
# 0 = Not Diabetic

diabetes_count = diabetes_df['Outcome'].sum()

print("Number of patients with diabetes:", diabetes_count)
```

```python
# Count how many patients have diabetes
# Assuming the Outcome column:
# 1 = Diabetic
# 0 = Not Diabetic

diabetes_count = diabetes_df['Outcome'].sum()

print("Number of patients with diabetes:", diabetes_count)

# Explanation
print("This makes sense because the Outcome column uses 1 to represent patients diagnosed with diabetes.")
```

```python
# Calculate the average glucose level

average_glucose = diabetes_df['Glucose'].mean()

print("Average glucose level:", average_glucose)

# Explanation
print("This makes sense because the mean function calculates the overall average glucose level of all patients in the dataset.")
```

```python
# Show only female patients above age 40
# (In the diabetes dataset, all patients are female)

female_above_40 = diabetes_df[diabetes_df['Age'] > 40]

print(female_above_40)

# Explanation
print("This makes sense because the filter selects only patients whose age is greater than 40.")
```

```python
# Group by Outcome and show average BMI

average_bmi = diabetes_df.groupby('Outcome')['BMI'].mean()

print(average_bmi)

# Explanation
print("This makes sense because grouping by Outcome helps compare the average BMI of diabetic and non-diabetic patients.")
```

```python
# Drop rows with missing or zero glucose values

original_rows = len(diabetes_df)

# Remove rows where Glucose is 0
df_cleaned = diabetes_df[diabetes_df['Glucose'] != 0]

new_rows = len(df_cleaned)

removed_rows = original_rows - new_rows

print("Number of rows removed:", removed_rows)

# Explanation
print("This makes sense because glucose values of 0 are usually invalid or missing in medical datasets.")
```

```python
# Import libraries
import matplotlib.pyplot as plt

# 1. Histogram
plt.figure(figsize=(6,4))
plt.hist(diabetes_df['Glucose'], bins=15)
plt.title('Histogram of Glucose Levels')
plt.xlabel('Glucose')
plt.ylabel('Frequency')
plt.show()

# Explanation
print("This chart makes sense because it shows how glucose values are distributed among patients.")
```

```python
# 2. Scatter Plot
plt.figure(figsize=(6,4))
plt.scatter(diabetes_df['Age'], diabetes_df['Glucose'])
plt.title('Age vs Glucose')
plt.xlabel('Age')
plt.ylabel('Glucose')
plt.show()

# Explanation
print("This chart makes sense because it helps visualize the relationship between age and glucose level.")
```

```python
# 3. Bar Chart
outcome_counts = diabetes_df['Outcome'].value_counts()

plt.figure(figsize=(6,4))
plt.bar(outcome_counts.index, outcome_counts.values)
plt.title('Diabetes Outcome Count')
plt.xlabel('Outcome')
plt.ylabel('Count')
plt.show()

# Explanation
print("This chart makes sense because it compares the number of diabetic and non-diabetic patients.")
```

```python
# 5. Line Chart
plt.figure(figsize=(8,4))
plt.plot(diabetes_df['Glucose'])
plt.title('Line Chart of Glucose Levels')
plt.xlabel('Patient Index')
plt.ylabel('Glucose')
plt.show()

# Explanation
print("This chart makes sense because it shows how glucose values change across the dataset.")
```

# Data Story Questions

1. Does glucose level affect diabetes outcome?

2. Which age group has the highest diabetes rate?

3. Is BMI linked to diabetes outcome?

```python
# Average glucose level by diabetes outcome

glucose_avg = diabetes_df.groupby('Outcome')['Glucose'].mean()

print(glucose_avg)

# Create bar chart
glucose_avg.plot(kind='bar')

plt.title('Average Glucose Level by Outcome')
plt.xlabel('Outcome')
plt.ylabel('Average Glucose')
plt.show()
```

### Finding

Patients with diabetes had a higher average glucose level compared to non-diabetic patients.

```python
# Create age groups

diabetes_df['AgeGroup'] = pd.cut(
    diabetes_df['Age'],
    bins=[20,30,40,50,60,70,80],
    labels=['21-30','31-40','41-50','51-60','61-70','71-80']
)

# Calculate diabetes rate by age group

age_group_diabetes = diabetes_df.groupby('AgeGroup')['Outcome'].mean()

print(age_group_diabetes)

# Create bar chart

age_group_diabetes.plot(kind='bar')

plt.title('Diabetes Rate by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Diabetes Rate')
plt.show()
```

### Finding

Older age groups showed a higher diabetes rate than younger age groups.

```python
# Calculate average BMI by diabetes outcome

bmi_avg = diabetes_df.groupby('Outcome')['BMI'].mean()

print(bmi_avg)

# Create bar chart

bmi_avg.plot(kind='bar')

plt.title('Average BMI by Outcome')
plt.xlabel('Outcome')
plt.ylabel('Average BMI')
plt.show()
```

### Finding

Patients with diabetes had a higher average BMI than patients without diabetes.
